**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# Write your code here for Q4

# Protected Variable (_): Isme ek underscore hota hai. Yeh sirf ek sign hai ke is variable ko bahar use na karein, lekin Python isko block nahi karta.
# Private Variable (__): Isme do underscore hote hain. Python isko strict block kar deta hai, isliye isko class ke bahar se direct access nahi kiya ja sakta.

class Mobile:
    def __init__(self, color, password):
        self._color = color #Protected variable (Single _)
        self.__password = password #Private variable (Double __)

    def get_password(self):
        print(f"Your password is {self.__password}")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def __init__(self, color):
        self.color = color

class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def show_circle(self):
        print(f"Circle Color: {self.color} | Radius: {self.radius}")

class Square(Shape):
    def __init__(self, color, side):
        super().__init__(color)
        self.side = side

    def show_square(self):
        print(f"Square Color: {self.color} | Side Length: {self.side}")
```

```python
c1 = Circle("Yellow",6)
s1 = Square("Red", 10)
```

```python
c1.show_circle()
s1.show_square()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Book:
    def __init__(self, pages):
        self.pages = pages

    def __add__(self, other):
        tot_pages = self.pages + other.pages
        return Book(tot_pages)
        
```

```python
b1 = Book(200) 
b2 = Book(300)
```

```python
b3 = b1 + b2
```

```python
print(b3.pages)
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

# 1. ABSTRACT PARENT CLASS (Payment System)

class Payment(ABC):
    def __init__(self, amount):
        self.amount = amount

    def show_amount(self):
        print(f" Total Billing Amount: RS: {self.amount}")

    @abstractmethod
    def process_payment(self):
        pass

# 2. CHILD CLASSES (Mandatory Implementation)

class CreditCardPayment(Payment):

     def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

     def process_payment(self):
        print(f" Processing Rs: {self.amount} securely via Credit Card ending in {self.card_number[-4:]}...")
        print(" Transaction Successful!")


class UPIPayment(Payment):
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def process_payment(self):
        print(f" Requesting Rs: {self.amount} from UPI ID: {self.upi_id}...")
        print("UPI Payment Received Successfully!")
```

```python
p1 = CreditCardPayment(1500, "1234567890123456")
p1.show_amount() 
p1.process_payment()
```

```python
p2 = UPIPayment(500, "hasan@ybl")
```

```python
p2.show_amount() 
```

```python
p2.process_payment()
```
